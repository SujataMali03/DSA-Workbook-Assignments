```python
# ============================================================
# ASSIGNMENT 1 - Q1
# ============================================================
#
# Question:
# Write a program that finds the largest and the smallest value
# in a list of numbers without using max() or min(). Report how
# many comparisons your code performed and state its time
# complexity in Big-O notation.
#
# Key Explanation:
# The program scans the list once and compares each element with
# the current largest and smallest values.
# max() and min() are not used.
#
# For each element after the first:
# 1. Compare it with the current largest value.
# 2. If it is not larger, compare it with the current smallest.
#
# Time Complexity: O(n)
# Space Complexity: O(1)
#
# The number of comparisons depends on the input values.


# Given list
numbers = [12, 5, 28, 7, 3, 19, 35, 10]

# Initialize largest and smallest
largest = numbers[0]
smallest = numbers[0]

# Variable to count comparisons
comparisons = 0

# Check every remaining element
for i in range(1, len(numbers)):

    # Compare with largest
    comparisons += 1

    if numbers[i] > largest:

        largest = numbers[i]

    else:

        # Compare with smallest
        comparisons += 1

        if numbers[i] < smallest:

            smallest = numbers[i]


# Display results
print("List:", numbers)
print("Largest value:", largest)
print("Smallest value:", smallest)
print("Number of comparisons:", comparisons)
print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

```python
# ============================================================
# ASSIGNMENT 1 - Q2
# ============================================================
#
# Question:
# Build a dictionary that classifies at least eight named data
# structures as linear or non-linear. Ask the user for a data
# structure name and report its type, then print all linear and
# all non-linear structures separately.
#
# Key Explanation:
# Linear data structures arrange elements sequentially.
# Non-linear data structures arrange elements hierarchically
# or as a network.
#
# The dictionary stores each data structure with its type.
# The user can enter a structure name to check its classification.
#
# ============================================================


# Dictionary containing data structures and their types

data_structures = {

    "array": "Linear",
    "linked list": "Linear",
    "stack": "Linear",
    "queue": "Linear",

    "tree": "Non-Linear",
    "binary tree": "Non-Linear",
    "heap": "Non-Linear",
    "graph": "Non-Linear"

}


# Ask the user for a data structure name

name = input(
    "Enter a data structure name: "
).strip().lower()


# Report the type of the entered structure

if name in data_structures:

    print(
        name.title(),
        "is a",
        data_structures[name],
        "data structure."
    )

else:

    print(
        "Data structure not found in the dictionary."
    )


# Print all linear data structures

print("\nLinear Data Structures:")

for structure, structure_type in data_structures.items():

    if structure_type == "Linear":

        print("-", structure.title())


# Print all non-linear data structures

print("\nNon-Linear Data Structures:")

for structure, structure_type in data_structures.items():

    if structure_type == "Non-Linear":

        print("-", structure.title())
```

```python
# ============================================================
# ASSIGNMENT 1 - Q3
# ============================================================
#
# Question:
# Investigate how running time grows with input size.
#
# a) Write three functions on a list of size n:
#    one O(1), one O(n) and one O(n²).
#
# b) Time each at n = 500, 1000 and 2000, and present the
#    results as a table of n against time.
#
# c) State which function grew fastest, and whether that
#    matches its Big-O complexity.
#
# Key Explanation:
# O(1) performs a constant amount of work.
# O(n) performs work proportional to the input size.
# O(n²) uses nested loops and performs approximately n²
# operations.
#
# The program uses time.perf_counter() to measure running time.
# The measured time can vary depending on the computer.
#
# ============================================================

import time


# ============================================================
# O(1) FUNCTION
# ============================================================

def constant_time(data):

    # Only one element is accessed.
    return data[0]


# ============================================================
# O(n) FUNCTION
# ============================================================

def linear_time(data):

    total = 0

    for value in data:

        total += value

    return total


# ============================================================
# O(n²) FUNCTION
# ============================================================

def quadratic_time(data):

    count = 0

    for i in range(len(data)):

        for j in range(len(data)):

            count += 1

    return count


# ============================================================
# FUNCTION TO MEASURE RUNNING TIME
# ============================================================

def measure_time(function, data):

    start_time = time.perf_counter()

    function(data)

    end_time = time.perf_counter()

    return end_time - start_time


# ============================================================
# INPUT SIZES
# ============================================================

sizes = [500, 1000, 2000]


# Store results
results = []


# ============================================================
# MEASURE EACH FUNCTION
# ============================================================

for n in sizes:

    data = list(range(n))

    constant_time_result = measure_time(
        constant_time,
        data
    )

    linear_time_result = measure_time(
        linear_time,
        data
    )

    quadratic_time_result = measure_time(
        quadratic_time,
        data
    )

    results.append(
        (
            n,
            constant_time_result,
            linear_time_result,
            quadratic_time_result
        )
    )


# ============================================================
# PART B - DISPLAY RESULTS AS A TABLE
# ============================================================

print(
    f"{'n':<10}"
    f"{'O(1) Time':<20}"
    f"{'O(n) Time':<20}"
    f"{'O(n²) Time':<20}"
)

print("-" * 70)


for row in results:

    n = row[0]
    constant_result = row[1]
    linear_result = row[2]
    quadratic_result = row[3]

    print(
        f"{n:<10}"
        f"{constant_result:<20.8f}"
        f"{linear_result:<20.8f}"
        f"{quadratic_result:<20.8f}"
    )


# ============================================================
# PART C - KEY OBSERVATION
# ============================================================

# The O(n²) function is expected to grow fastest as n increases.
#
# This matches its Big-O complexity because the number of
# operations increases approximately with n².
#
# O(1) grows very slowly because its work remains constant.
#
# O(n) grows approximately in proportion to n.
#
# O(n²) grows much faster as the input size increases.
#
# Note:
# Actual measured running times may vary slightly between
# executions because of computer/system conditions.
```

```python
# ============================================================
# ASSIGNMENT 1 - Q4
# ============================================================
#
# Question:
# Count the comparisons a linear search makes on a list of
# 1000 elements for:
# 1. A key that is the first element
# 2. A key in the middle
# 3. A key that is absent
#
# Identify which of the three is the best case and which is
# the worst case, and explain why.
#
# Key Explanation:
# Linear search checks elements one by one from the beginning.
#
# If the key is the first element, only 1 comparison is needed.
# If the key is in the middle, 501 comparisons are needed for
# index 500.
# If the key is absent, all 1000 elements must be checked.
#
# Best Case: Key is the first element -> 1 comparison.
# Worst Case: Key is absent -> 1000 comparisons.
#
# Linear Search Time Complexity:
# Best Case  -> O(1)
# Worst Case -> O(n)
#
# ============================================================


# Create a list containing 1000 elements

numbers = list(range(1, 1001))


# ------------------------------------------------------------
# Linear Search Function
# ------------------------------------------------------------

def linear_search(numbers, key):

    comparisons = 0

    for i in range(len(numbers)):

        comparisons += 1

        if numbers[i] == key:

            return i, comparisons

    return -1, comparisons


# ------------------------------------------------------------
# Case 1: Key is the first element
# ------------------------------------------------------------

first_key = numbers[0]

first_index, first_comparisons = linear_search(
    numbers,
    first_key
)


# ------------------------------------------------------------
# Case 2: Key is in the middle
# ------------------------------------------------------------

middle_key = numbers[500]

middle_index, middle_comparisons = linear_search(
    numbers,
    middle_key
)


# ------------------------------------------------------------
# Case 3: Key is absent
# ------------------------------------------------------------

absent_key = 2000

absent_index, absent_comparisons = linear_search(
    numbers,
    absent_key
)


# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("Number of elements:", len(numbers))

print("\nCase 1: Key is the first element")
print("Key:", first_key)
print("Comparisons:", first_comparisons)

print("\nCase 2: Key is in the middle")
print("Key:", middle_key)
print("Comparisons:", middle_comparisons)

print("\nCase 3: Key is absent")
print("Key:", absent_key)
print("Comparisons:", absent_comparisons)


# ------------------------------------------------------------
# Best and Worst Case
# ------------------------------------------------------------

print("\nBest Case:")
print("Key is the first element.")
print("Number of comparisons:", first_comparisons)

print("\nWorst Case:")
print("Key is absent.")
print("Number of comparisons:", absent_comparisons)


# ------------------------------------------------------------
# Key Results
# ------------------------------------------------------------
#
# First element:
# 1 comparison
#
# Middle element:
# 501 comparisons
#
# Absent key:
# 1000 comparisons
#
# Best Case = First element
# Worst Case = Absent key
#
# Reason:
# The first-element case stops immediately, while an absent key
# requires checking every element in the list.
#
# Linear search has O(1) best-case complexity and O(n)
# worst-case complexity.
```

```python
# ============================================================
# ASSIGNMENT 1 - Q5
# ============================================================
#
# Question:
# Compute the sum 1 + 2 + … + n in two ways:
# 1. With a loop
# 2. With the formula n × (n + 1) / 2
#
# Verify that both give the same answer for n = 1 to 100,
# time both at n = 1000000, and state the complexity of each.
# Explain what this shows about the value of choosing a
# better algorithm.
#
# Key Explanation:
# The loop adds every number from 1 to n, so it performs n
# additions and has O(n) time complexity.
#
# The formula directly calculates the sum using one expression,
# so it has O(1) time complexity.
#
# Both methods should produce the same answer.
# The formula becomes much more efficient for large values of n.
#
# ============================================================

import time


# ============================================================
# METHOD 1 - USING A LOOP
# ============================================================

def sum_using_loop(n):

    total = 0

    for i in range(1, n + 1):

        total += i

    return total


# ============================================================
# METHOD 2 - USING FORMULA
# ============================================================

def sum_using_formula(n):

    return n * (n + 1) // 2


# ============================================================
# PART 1 - VERIFY FOR n = 1 TO 100
# ============================================================

all_same = True

for n in range(1, 101):

    loop_sum = sum_using_loop(n)

    formula_sum = sum_using_formula(n)

    if loop_sum != formula_sum:

        all_same = False

        print(
            "Different result found for n =",
            n
        )


print("Verification for n = 1 to 100:")

if all_same:

    print(
        "Both methods give the same answer for all values."
    )

else:

    print(
        "The results are not the same."
    )


# ============================================================
# PART 2 - DISPLAY EXAMPLE
# ============================================================

print("\nExample for n = 100:")

print(
    "Sum using loop:",
    sum_using_loop(100)
)

print(
    "Sum using formula:",
    sum_using_formula(100)
)


# ============================================================
# PART 3 - TIME AT n = 1000000
# ============================================================

n = 1000000


# Time loop method

start_time = time.perf_counter()

loop_result = sum_using_loop(n)

loop_time = time.perf_counter() - start_time


# Time formula method

start_time = time.perf_counter()

formula_result = sum_using_formula(n)

formula_time = time.perf_counter() - start_time


# ============================================================
# PART 4 - DISPLAY TIMING RESULTS
# ============================================================

print("\nTiming for n = 1000000")

print(
    "Loop result:",
    loop_result
)

print(
    "Loop time:",
    loop_time,
    "seconds"
)

print(
    "\nFormula result:",
    formula_result
)

print(
    "Formula time:",
    formula_time,
    "seconds"
)


# ============================================================
# PART 5 - VERIFY BOTH RESULTS
# ============================================================

print("\nFinal Verification:")

if loop_result == formula_result:

    print(
        "Both methods produce the same answer."
    )

else:

    print(
        "The methods produce different answers."
    )


# ============================================================
# PART 6 - COMPLEXITY
# ============================================================

# Loop Method:
# Time Complexity = O(n)
# Space Complexity = O(1)
#
# Formula Method:
# Time Complexity = O(1)
# Space Complexity = O(1)
#
# The loop performs n additions.
# The formula calculates the result directly.
#
# Therefore, the formula is more efficient for very large n.
#
# This demonstrates the importance of choosing a better algorithm.
# Two algorithms can solve the same problem and produce the same
# answer, but their running times can be very different.
#
# A better algorithm can reduce the amount of work required,
# especially when the input size becomes very large.
```
