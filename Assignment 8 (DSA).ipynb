```python
# Assignment 8 - Q1
#
# Question:
# Implement a doubly linked list supporting:
# a) insert_at_head, insert_at_tail and insert_after a given node
# b) delete_by_value, correctly handling head, tail and middle nodes
# c) forward traversal and backward traversal
# Also test the operations on an empty list, a one-element list,
# and deletion of the only node.
#
# Key Explanation:
# A doubly linked list node contains data, a previous reference
# and a next reference.
# The previous and next references allow traversal in both directions.
# Head points to the first node and tail points to the last node.

class DNode:

    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    # Insert at head
    def insert_at_head(self, data):

        node = DNode(data)

        if self.head is None:
            self.head = self.tail = node
            return

        node.next = self.head
        self.head.prev = node
        self.head = node

    # Insert at tail
    def insert_at_tail(self, data):

        node = DNode(data)

        if self.tail is None:
            self.head = self.tail = node
            return

        node.prev = self.tail
        self.tail.next = node
        self.tail = node

    # Insert after a given node
    def insert_after(self, node, data):

        if node is None:
            return

        new_node = DNode(data)

        new_node.prev = node
        new_node.next = node.next

        if node.next is not None:
            node.next.prev = new_node
        else:
            self.tail = new_node

        node.next = new_node

    # Delete by value
    def delete_by_value(self, data):

        current = self.head

        while current is not None:

            if current.data == data:

                if current.prev is None:
                    self.head = current.next
                else:
                    current.prev.next = current.next

                if current.next is None:
                    self.tail = current.prev
                else:
                    current.next.prev = current.prev

                return True

            current = current.next

        return False

    # Forward traversal
    def forward(self):

        values = []
        current = self.head

        while current is not None:
            values.append(str(current.data))
            current = current.next

        return " <-> ".join(values)

    # Backward traversal
    def backward(self):

        values = []
        current = self.tail

        while current is not None:
            values.append(str(current.data))
            current = current.prev

        return " <-> ".join(values)


# Test on an empty list
dll = DoublyLinkedList()

print("Empty list:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Test on a one-element list
dll.insert_at_head(10)

print("\nOne-element list:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Insert at head and tail
dll.insert_at_head(5)
dll.insert_at_tail(20)

print("\nAfter inserting at head and tail:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Insert after a given node
node = dll.head.next
dll.insert_after(node, 15)

print("\nAfter inserting 15:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Delete head
dll.delete_by_value(5)

print("\nAfter deleting head:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Delete middle node
dll.delete_by_value(15)

print("\nAfter deleting middle node:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Delete tail
dll.delete_by_value(20)

print("\nAfter deleting tail:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Delete the only remaining node
dll.delete_by_value(10)

print("\nAfter deleting the only node:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())
```

```python
# Assignment 8 - Q2
#
# Question:
# Build a doubly linked list of 10 elements and print it forwards
# and then backwards, confirming that the second output is the
# reverse of the first.
# Then write a search method that reports both the position of
# a value counted from the head and its position counted from the tail.
#
# Key Explanation:
# A doubly linked list has both next and prev references.
# Forward traversal starts from the head, while backward traversal
# starts from the tail.
# The position from the tail can be found by traversing backward.

class DNode:

    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    # Insert at tail
    def insert_at_tail(self, data):

        node = DNode(data)

        if self.head is None:
            self.head = self.tail = node
            return

        node.prev = self.tail
        self.tail.next = node
        self.tail = node

    # Forward traversal
    def forward(self):

        values = []
        current = self.head

        while current is not None:
            values.append(current.data)
            current = current.next

        return values

    # Backward traversal
    def backward(self):

        values = []
        current = self.tail

        while current is not None:
            values.append(current.data)
            current = current.prev

        return values

    # Search and report position from head and tail
    def search(self, value):

        # Position counted from head
        current = self.head
        position_from_head = 0

        while current is not None:

            if current.data == value:
                break

            current = current.next
            position_from_head += 1

        # Value not found
        if current is None:
            return -1, -1

        # Position counted from tail
        current = self.tail
        position_from_tail = 0

        while current is not None:

            if current.data == value:
                break

            current = current.prev
            position_from_tail += 1

        return position_from_head, position_from_tail


# --------------------------------------------------
# Build a doubly linked list of 10 elements
# --------------------------------------------------

dll = DoublyLinkedList()

for i in range(1, 11):
    dll.insert_at_tail(i)


# --------------------------------------------------
# Print forwards and backwards
# --------------------------------------------------

forward_list = dll.forward()
backward_list = dll.backward()

print("Forward:", forward_list)
print("Backward:", backward_list)


# Confirm backward output is the reverse of forward output

if backward_list == forward_list[::-1]:
    print("Backward traversal is the reverse of forward traversal.")


# --------------------------------------------------
# Search for a value
# --------------------------------------------------

value = 7

head_position, tail_position = dll.search(value)

print("\nValue:", value)
print("Position from head:", head_position)
print("Position from tail:", tail_position)
```

```python
# Assignment 8 - Q3
#
# Question:
# Build a doubly linked list of 20 elements.
# Delete the head, the tail and a middle node,
# printing the list in both directions after each deletion.
# Draw the references before and after the middle deletion by hand
# and attach the drawing.
#
# Key Explanation:
# A doubly linked list contains both prev and next references.
# When a node is deleted, the previous node's next reference
# and the next node's prev reference must be updated.
# Head and tail are also updated when the first or last node is deleted.

class DNode:

    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    # Insert at tail
    def insert_at_tail(self, data):

        node = DNode(data)

        if self.head is None:
            self.head = self.tail = node
            return

        node.prev = self.tail
        self.tail.next = node
        self.tail = node

    # Delete by value
    def delete_by_value(self, data):

        current = self.head

        while current is not None:

            if current.data == data:

                # If deleting the head
                if current.prev is None:
                    self.head = current.next
                else:
                    current.prev.next = current.next

                # If deleting the tail
                if current.next is None:
                    self.tail = current.prev
                else:
                    current.next.prev = current.prev

                return True

            current = current.next

        return False

    # Forward traversal
    def forward(self):

        values = []
        current = self.head

        while current is not None:
            values.append(str(current.data))
            current = current.next

        return " <-> ".join(values)

    # Backward traversal
    def backward(self):

        values = []
        current = self.tail

        while current is not None:
            values.append(str(current.data))
            current = current.prev

        return " <-> ".join(values)


# --------------------------------------------------
# Build a doubly linked list of 20 elements
# --------------------------------------------------

dll = DoublyLinkedList()

for i in range(1, 21):
    dll.insert_at_tail(i)


# --------------------------------------------------
# Original list
# --------------------------------------------------

print("Original list:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# --------------------------------------------------
# Delete the head
# --------------------------------------------------

dll.delete_by_value(1)

print("\nAfter deleting head:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# --------------------------------------------------
# Delete the tail
# --------------------------------------------------

dll.delete_by_value(20)

print("\nAfter deleting tail:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# --------------------------------------------------
# Delete a middle node
# --------------------------------------------------

dll.delete_by_value(10)

print("\nAfter deleting middle node:")
print("Forward:", dll.forward())
print("Backward:", dll.backward())


# Reference changes for middle deletion:
#
# Before:
# Node 9 <-> Node 10 <-> Node 11
#
# After:
# Node 9 <-> Node 11
#
# Node 9's next is changed to Node 11.
# Node 11's prev is changed to Node 9.
```

```python
# Assignment 8 - Q4
#
# Question:
# Use your doubly linked list to implement a simple browser history
# supporting visit(page), back() and forward().
# Demonstrate it with at least eight operations, and explain why
# a doubly linked list suits this problem better than a singly linked one.
#
# Key Explanation:
# The current node represents the current web page.
# The prev reference moves backward in history and the next reference
# moves forward in history.
# A doubly linked list is suitable because both back and forward
# navigation are required.


class PageNode:

    def __init__(self, page):
        self.page = page
        self.prev = None
        self.next = None


class BrowserHistory:

    def __init__(self):
        self.current = None

    # Visit a new page
    def visit(self, page):

        new_node = PageNode(page)

        if self.current is None:
            self.current = new_node
            return

        # Remove forward history
        self.current.next = None

        new_node.prev = self.current
        self.current.next = new_node

        self.current = new_node

    # Move one page backward
    def back(self):

        if self.current is not None and self.current.prev is not None:
            self.current = self.current.prev

        return self.current.page

    # Move one page forward
    def forward(self):

        if self.current is not None and self.current.next is not None:
            self.current = self.current.next

        return self.current.page

    # Return current page
    def current_page(self):

        if self.current is None:
            return None

        return self.current.page


# --------------------------------------------------
# Browser history demonstration
# --------------------------------------------------

history = BrowserHistory()


# Operation 1
history.visit("google.com")
print("1. Visit:", history.current_page())


# Operation 2
history.visit("youtube.com")
print("2. Visit:", history.current_page())


# Operation 3
history.visit("github.com")
print("3. Visit:", history.current_page())


# Operation 4
history.back()
print("4. Back:", history.current_page())


# Operation 5
history.back()
print("5. Back:", history.current_page())


# Operation 6
history.forward()
print("6. Forward:", history.current_page())


# Operation 7
history.visit("wikipedia.org")
print("7. Visit:", history.current_page())


# Operation 8
history.back()
print("8. Back:", history.current_page())


# Explanation:
# A doubly linked list is better suited because each page has
# both previous and next references.
# Therefore, moving backward and forward can be done directly.
# In a singly linked list, forward movement is possible, but
# moving backward requires finding the previous node again.
```

```python
# Assignment 8 - Q5
#
# Question:
# Compare deletion in the two kinds of list.
# Implement a method that deletes a node you already hold
# a reference to, for both a singly and a doubly linked list
# of 5000 elements.
# Measure both, and explain why one is O(1) while the other is O(n).
#
# Key Explanation:
# In a singly linked list, even when the target node is known,
# its predecessor must be found before changing the next reference.
# Therefore, deletion is O(n).
#
# In a doubly linked list, the node already has a prev reference,
# so its predecessor is directly available. Therefore, deletion
# is O(1).


import time


# --------------------------------------------------
# Singly Linked List
# --------------------------------------------------

class SNode:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_tail(self, data):

        node = SNode(data)

        if self.head is None:
            self.head = self.tail = node
            return

        self.tail.next = node
        self.tail = node

    # Delete a node when its reference is already known
    def delete_node(self, target):

        if target is None or self.head is None:
            return False

        # If target is the head
        if target == self.head:
            self.head = self.head.next

            if self.head is None:
                self.tail = None

            return True

        # Find the predecessor of target
        previous = self.head

        while previous.next is not None:
            if previous.next == target:

                previous.next = target.next

                if target == self.tail:
                    self.tail = previous

                return True

            previous = previous.next

        return False


# --------------------------------------------------
# Doubly Linked List
# --------------------------------------------------

class DNode:

    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_tail(self, data):

        node = DNode(data)

        if self.head is None:
            self.head = self.tail = node
            return

        node.prev = self.tail
        self.tail.next = node
        self.tail = node

    # Delete a node when its reference is already known
    def delete_node(self, target):

        if target is None:
            return False

        # Update previous node
        if target.prev is not None:
            target.prev.next = target.next
        else:
            self.head = target.next

        # Update next node
        if target.next is not None:
            target.next.prev = target.prev
        else:
            self.tail = target.prev

        target.prev = None
        target.next = None

        return True


# --------------------------------------------------
# Create both lists with 5000 elements
# --------------------------------------------------

singly_list = SinglyLinkedList()
doubly_list = DoublyLinkedList()

for i in range(5000):
    singly_list.insert_at_tail(i)
    doubly_list.insert_at_tail(i)


# --------------------------------------------------
# Hold a reference to the middle node
# --------------------------------------------------

singly_target = singly_list.head

for _ in range(2500):
    singly_target = singly_target.next


doubly_target = doubly_list.head

for _ in range(2500):
    doubly_target = doubly_target.next


# --------------------------------------------------
# Measure singly linked list deletion
# --------------------------------------------------

start = time.perf_counter()

singly_list.delete_node(singly_target)

singly_time = time.perf_counter() - start


# --------------------------------------------------
# Measure doubly linked list deletion
# --------------------------------------------------

start = time.perf_counter()

doubly_list.delete_node(doubly_target)

doubly_time = time.perf_counter() - start


# --------------------------------------------------
# Display measured times
# --------------------------------------------------

print("Singly linked list deletion time:",
      singly_time, "seconds")

print("Doubly linked list deletion time:",
      doubly_time, "seconds")


# --------------------------------------------------
# Complexity explanation
# --------------------------------------------------

# Singly linked list:
# O(n), because the predecessor of the target node
# must be found by traversing from the head.
#
# Doubly linked list:
# O(1), because target.prev directly gives the predecessor
# and target.next directly gives the successor.
#
# The workbook states:
# Singly linked: deleting a held node is O(n).
# Doubly linked: deleting a held node is O(1).
```
