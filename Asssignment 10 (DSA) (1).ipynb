```python
# ============================================================
# ASSIGNMENT 10 - FINAL PROJECT
# TICKET BOOKING QUEUE
# ============================================================

# ------------------------------------------------------------
# QUESTION / PROJECT REQUIREMENT
# ------------------------------------------------------------
#
# Choose the Ticket Booking Queue project.
#
# Simulate a counter that serves customers in arrival order,
# including cancellations. Compare a circular queue with a
# linked-list queue, considering that cancellation may remove
# someone from the middle.
#
# Required:
# 1. Design note before coding.
# 2. Working Python implementation.
# 3. Test driver.
# 4. Performance measurement using at least 10,000 records
#    or an equivalent workload.
# 5. Identify a bottleneck.
# 6. Improve the solution.
# 7. Provide before-and-after measurements.
# 8. Prepare a short report.
#
# Key Explanation:
# A queue follows FIFO (First In, First Out).
# The customer who arrives first is served first.
#
# Two implementations are used:
# 1. Circular Queue - array-based implementation.
# 2. Linked List Queue - node-based implementation.
#
# Cancellation is also supported because a customer may cancel
# a booking before being served.


# ============================================================
# PART 1 - DESIGN NOTE
# ============================================================

# Problem:
# A ticket counter needs to manage customers waiting for tickets.
# Customers should be served in their arrival order.
# A customer may also cancel while waiting.
#
# Data Structure 1:
# Circular Queue
# - Uses a fixed-size array.
# - Uses front and rear positions.
# - Reuses empty positions using circular movement.
#
# Data Structure 2:
# Linked List Queue
# - Uses dynamically allocated nodes.
# - Uses front and rear references.
# - Does not require a fixed array capacity.
#
# Main Operations:
# - enqueue() -> Add a customer.
# - dequeue() -> Serve the first customer.
# - cancel() -> Cancel a customer's booking.
# - display() -> Display waiting customers.
#
# Complexity:
# Circular Queue enqueue  -> O(1)
# Circular Queue dequeue  -> O(1)
# Circular Queue cancellation -> O(n)
#
# Linked List enqueue -> O(1)
# Linked List dequeue -> O(1)
# Linked List cancellation -> O(n) when searching by customer ID.
#
# The main bottleneck is cancellation from the middle of the
# circular queue.


# ============================================================
# PART 2 - CIRCULAR QUEUE
# ============================================================

class CircularTicketQueue:

    def __init__(self, capacity):

        self.capacity = capacity
        self.queue = [None] * capacity

        self.front = 0
        self.rear = 0
        self.size = 0

    # --------------------------------------------------------
    # Check whether queue is empty
    # --------------------------------------------------------

    def is_empty(self):

        return self.size == 0

    # --------------------------------------------------------
    # Check whether queue is full
    # --------------------------------------------------------

    def is_full(self):

        return self.size == self.capacity

    # --------------------------------------------------------
    # Add customer
    # --------------------------------------------------------

    def enqueue(self, customer):

        if self.is_full():

            return False

        self.queue[self.rear] = customer

        self.rear = (
            self.rear + 1
        ) % self.capacity

        self.size += 1

        return True

    # --------------------------------------------------------
    # Serve first customer
    # --------------------------------------------------------

    def dequeue(self):

        if self.is_empty():

            return None

        customer = self.queue[self.front]

        self.queue[self.front] = None

        self.front = (
            self.front + 1
        ) % self.capacity

        self.size -= 1

        return customer

    # --------------------------------------------------------
    # Cancel customer
    # --------------------------------------------------------

    def cancel(self, customer_id):

        if self.is_empty():

            return False

        customers = []

        current = self.front

        # Copy active customers into a temporary list
        for _ in range(self.size):

            customers.append(
                self.queue[current]
            )

            current = (
                current + 1
            ) % self.capacity

        # Remove requested customer
        new_customers = [

            customer
            for customer in customers
            if customer["id"] != customer_id

        ]

        # Customer was not found
        if len(new_customers) == len(customers):

            return False

        # Rebuild queue
        self.queue = [None] * self.capacity

        self.front = 0
        self.rear = 0
        self.size = 0

        for customer in new_customers:

            self.enqueue(customer)

        return True

    # --------------------------------------------------------
    # Display queue
    # --------------------------------------------------------

    def display(self):

        if self.is_empty():

            print("Queue is empty.")

            return

        current = self.front

        print("\nCircular Queue:")

        for position in range(1, self.size + 1):

            customer = self.queue[current]

            print(
                position,
                "->",
                customer["id"],
                customer["name"]
            )

            current = (
                current + 1
            ) % self.capacity


# ============================================================
# PART 3 - LINKED LIST QUEUE
# ============================================================

class TicketNode:

    def __init__(self, customer):

        self.customer = customer
        self.next = None


class LinkedListTicketQueue:

    def __init__(self):

        self.front = None
        self.rear = None

    # --------------------------------------------------------
    # Check whether queue is empty
    # --------------------------------------------------------

    def is_empty(self):

        return self.front is None

    # --------------------------------------------------------
    # Add customer
    # --------------------------------------------------------

    def enqueue(self, customer):

        new_node = TicketNode(customer)

        if self.front is None:

            self.front = new_node
            self.rear = new_node

            return True

        self.rear.next = new_node

        self.rear = new_node

        return True

    # --------------------------------------------------------
    # Serve first customer
    # --------------------------------------------------------

    def dequeue(self):

        if self.is_empty():

            return None

        customer = self.front.customer

        self.front = self.front.next

        if self.front is None:

            self.rear = None

        return customer

    # --------------------------------------------------------
    # Cancel customer
    # --------------------------------------------------------

    def cancel(self, customer_id):

        if self.is_empty():

            return False

        current = self.front
        previous = None

        while current is not None:

            if current.customer["id"] == customer_id:

                # Delete first node
                if previous is None:

                    self.front = current.next

                # Delete middle or last node
                else:

                    previous.next = current.next

                # Update rear if required
                if current == self.rear:

                    self.rear = previous

                return True

            previous = current
            current = current.next

        return False

    # --------------------------------------------------------
    # Display queue
    # --------------------------------------------------------

    def display(self):

        if self.is_empty():

            print("Queue is empty.")

            return

        current = self.front
        position = 1

        print("\nLinked List Queue:")

        while current is not None:

            customer = current.customer

            print(
                position,
                "->",
                customer["id"],
                customer["name"]
            )

            current = current.next

            position += 1


# ============================================================
# PART 4 - CUSTOMER DATA
# ============================================================

def create_customer(customer_id, name):

    return {
        "id": customer_id,
        "name": name
    }


customers = [

    create_customer(101, "Aarav"),
    create_customer(102, "Priya"),
    create_customer(103, "Rahul"),
    create_customer(104, "Sneha"),
    create_customer(105, "Rohan"),
    create_customer(106, "Ananya"),
    create_customer(107, "Vikas"),
    create_customer(108, "Neha")

]


# ============================================================
# PART 5 - CREATE BOTH QUEUES
# ============================================================

circular_queue = CircularTicketQueue(20)

linked_queue = LinkedListTicketQueue()


# ============================================================
# PART 6 - ADD CUSTOMERS
# ============================================================

for customer in customers:

    circular_queue.enqueue(customer)

    linked_queue.enqueue(customer)


print("INITIAL QUEUES")

circular_queue.display()

linked_queue.display()


# ============================================================
# PART 7 - MIDDLE CANCELLATION
# ============================================================

customer_to_cancel = 104

print("\nCANCELLING CUSTOMER:", customer_to_cancel)

circular_result = circular_queue.cancel(
    customer_to_cancel
)

linked_result = linked_queue.cancel(
    customer_to_cancel
)

print(
    "Circular Queue cancellation:",
    circular_result
)

print(
    "Linked List Queue cancellation:",
    linked_result
)


# ============================================================
# PART 8 - DISPLAY AFTER CANCELLATION
# ============================================================

print("\nAFTER CANCELLATION")

circular_queue.display()

linked_queue.display()


# ============================================================
# PART 9 - SERVE CUSTOMERS
# ============================================================

print("\nSERVING CUSTOMERS FROM CIRCULAR QUEUE")

while not circular_queue.is_empty():

    customer = circular_queue.dequeue()

    print(
        "Served:",
        customer["id"],
        customer["name"]
    )


print("\nSERVING CUSTOMERS FROM LINKED LIST QUEUE")

while not linked_queue.is_empty():

    customer = linked_queue.dequeue()

    print(
        "Served:",
        customer["id"],
        customer["name"]
    )


# ============================================================
# PART 10 - LARGE DATASET OF 10,000 CUSTOMERS
# ============================================================

import time


large_customers = [

    create_customer(
        i,
        "Customer_" + str(i)
    )

    for i in range(1, 10001)

]


# ============================================================
# PART 11 - CIRCULAR QUEUE PERFORMANCE
# ============================================================

large_circular_queue = CircularTicketQueue(10000)


start_time = time.perf_counter()


for customer in large_customers:

    large_circular_queue.enqueue(customer)


circular_enqueue_time = (
    time.perf_counter() - start_time
)


# Measure middle cancellation

start_time = time.perf_counter()


large_circular_queue.cancel(5000)


circular_cancel_time = (
    time.perf_counter() - start_time
)


# ============================================================
# PART 12 - LINKED LIST PERFORMANCE
# ============================================================

large_linked_queue = LinkedListTicketQueue()


start_time = time.perf_counter()


for customer in large_customers:

    large_linked_queue.enqueue(customer)


linked_enqueue_time = (
    time.perf_counter() - start_time
)


# Measure middle cancellation

start_time = time.perf_counter()


large_linked_queue.cancel(5000)


linked_cancel_time = (
    time.perf_counter() - start_time
)


# ============================================================
# PART 13 - PERFORMANCE RESULTS
# ============================================================

print("\n==========================================")
print("PERFORMANCE RESULTS")
print("==========================================")

print(
    "\nNumber of customers:",
    len(large_customers)
)

print(
    "\nCircular Queue Enqueue:",
    circular_enqueue_time,
    "seconds"
)

print(
    "Circular Queue Cancellation:",
    circular_cancel_time,
    "seconds"
)

print(
    "\nLinked List Enqueue:",
    linked_enqueue_time,
    "seconds"
)

print(
    "Linked List Cancellation:",
    linked_cancel_time,
    "seconds"
)


# ============================================================
# PART 14 - BOTTLENECK
# ============================================================

# Key Explanation:
#
# The main bottleneck in the original circular queue is
# cancellation from the middle.
#
# The original method:
# 1. Copies all active customers into another list.
# 2. Removes the cancelled customer.
# 3. Clears the queue.
# 4. Inserts all remaining customers again.
#
# This causes unnecessary copying and rebuilding.
#
# The time complexity remains O(n), but the implementation
# performs extra work.
#
# Improvement:
# Shift only the elements after the cancelled customer
# directly inside the circular array.
#
# This avoids creating a temporary list and rebuilding
# the entire queue.


# ============================================================
# PART 15 - IMPROVED CIRCULAR QUEUE
# ============================================================

class ImprovedCircularTicketQueue:

    def __init__(self, capacity):

        self.capacity = capacity
        self.queue = [None] * capacity

        self.front = 0
        self.rear = 0
        self.size = 0

    def is_empty(self):

        return self.size == 0

    def is_full(self):

        return self.size == self.capacity

    def enqueue(self, customer):

        if self.is_full():

            return False

        self.queue[self.rear] = customer

        self.rear = (
            self.rear + 1
        ) % self.capacity

        self.size += 1

        return True

    def dequeue(self):

        if self.is_empty():

            return None

        customer = self.queue[self.front]

        self.queue[self.front] = None

        self.front = (
            self.front + 1
        ) % self.capacity

        self.size -= 1

        return customer

    # Improved cancellation
    def cancel(self, customer_id):

        if self.is_empty():

            return False

        found_position = -1

        # Find customer position
        for i in range(self.size):

            index = (
                self.front + i
            ) % self.capacity

            if self.queue[index]["id"] == customer_id:

                found_position = i

                break

        # Customer not found
        if found_position == -1:

            return False

        # Shift only required elements
        for i in range(
            found_position,
            self.size - 1
        ):

            current_index = (
                self.front + i
            ) % self.capacity

            next_index = (
                self.front + i + 1
            ) % self.capacity

            self.queue[current_index] = (
                self.queue[next_index]
            )

        # Move rear backward
        self.rear = (
            self.rear - 1
        ) % self.capacity

        self.queue[self.rear] = None

        self.size -= 1

        return True

    def display(self):

        if self.is_empty():

            print("Queue is empty.")

            return

        current = self.front

        print("\nImproved Circular Queue:")

        for position in range(1, self.size + 1):

            customer = self.queue[current]

            print(
                position,
                "->",
                customer["id"],
                customer["name"]
            )

            current = (
                current + 1
            ) % self.capacity


# ============================================================
# PART 16 - BEFORE MEASUREMENT
# ============================================================

original_queue = CircularTicketQueue(10000)


for customer in large_customers:

    original_queue.enqueue(customer)


start_time = time.perf_counter()


original_queue.cancel(5000)


before_time = (
    time.perf_counter() - start_time
)


# ============================================================
# PART 17 - AFTER MEASUREMENT
# ============================================================

improved_queue = ImprovedCircularTicketQueue(10000)


for customer in large_customers:

    improved_queue.enqueue(customer)


start_time = time.perf_counter()


improved_queue.cancel(5000)


after_time = (
    time.perf_counter() - start_time
)


# ============================================================
# PART 18 - BEFORE / AFTER COMPARISON
# ============================================================

print("\n==========================================")
print("BEFORE / AFTER IMPROVEMENT")
print("==========================================")

print(
    "\nBefore improvement:",
    before_time,
    "seconds"
)

print(
    "After improvement:",
    after_time,
    "seconds"
)

if after_time < before_time:

    print(
        "\nThe improved implementation was faster "
        "in this execution."
    )

elif after_time > before_time:

    print(
        "\nThe improved implementation was slower "
        "in this execution."
    )

else:

    print(
        "\nBoth implementations took approximately "
        "the same time."
    )


# ============================================================
# PART 19 - COMPLEXITY COMPARISON
# ============================================================

# Circular Queue:
#
# Enqueue:
# O(1)
#
# Dequeue:
# O(1)
#
# Original cancellation:
# O(n)
#
# Improved cancellation:
# O(n)
#
# The improved version keeps the same asymptotic complexity
# but avoids temporary-list creation and complete queue rebuilding.
#
#
# Linked List Queue:
#
# Enqueue:
# O(1)
#
# Dequeue:
# O(1)
#
# Cancellation by customer ID:
# O(n)
#
# Once the required node and its predecessor are known,
# the actual link removal is O(1).


# ============================================================
# PART 20 - FINAL TEST
# ============================================================

print("\n==========================================")
print("FINAL TEST")
print("==========================================")


test_circular = CircularTicketQueue(5)

test_linked = LinkedListTicketQueue()


test_customers = [

    create_customer(1, "Customer A"),
    create_customer(2, "Customer B"),
    create_customer(3, "Customer C"),
    create_customer(4, "Customer D"),
    create_customer(5, "Customer E")

]


for customer in test_customers:

    test_circular.enqueue(customer)

    test_linked.enqueue(customer)


# Cancel middle customer

test_circular.cancel(3)

test_linked.cancel(3)


print("\nAfter cancelling Customer 3:")

test_circular.display()

test_linked.display()


# Serve remaining customers

print("\nServing Circular Queue:")

while not test_circular.is_empty():

    customer = test_circular.dequeue()

    print(
        customer["id"],
        customer["name"]
    )


print("\nServing Linked List Queue:")

while not test_linked.is_empty():

    customer = test_linked.dequeue()

    print(
        customer["id"],
        customer["name"]
    )


# ============================================================
# PART 21 - FINAL REPORT
# ============================================================

# FINAL REPORT
#
# Title: Ticket Booking Queue
#
# Introduction:
# The Ticket Booking Queue project demonstrates the application
# of queue data structures in a real-world ticket booking system.
# Customers arrive at a ticket counter and wait to be served.
# Since customers should normally be served according to their
# arrival order, the FIFO (First In, First Out) principle is used.
# The project implements the queue using both a circular queue
# and a linked-list queue.
#
# Implementation:
# The circular queue uses a fixed-size array with front and rear
# positions. Circular movement allows positions that become empty
# after dequeue operations to be reused. The linked-list queue
# uses nodes containing customer information and a reference to
# the next node. Front and rear references are maintained for
# efficient insertion and deletion.
#
# Both implementations support customer arrival, serving,
# cancellation and display operations. Cancellation is important
# because a customer may cancel a ticket while waiting. The project
# specifically tests cancellation from the middle of the queue.
#
# Performance Testing:
# A workload of 10,000 customer records was used for performance
# measurement. Python's time.perf_counter() function was used to
# measure enqueue and cancellation operations.
#
# The circular queue provides O(1) enqueue and dequeue operations.
# However, cancellation by customer ID requires searching through
# the queue and therefore takes O(n) time.
#
# The linked-list queue also requires O(n) time when a customer
# must first be located by ID. After the required node is found,
# the links can be changed to remove the node without shifting
# other customers.
#
# Bottleneck and Improvement:
# The main bottleneck identified in the original circular queue
# was middle cancellation. The original implementation created
# a temporary list and rebuilt the complete queue after removing
# a customer. This caused unnecessary copying and rebuilding.
#
# An improved circular queue was implemented that shifts only
# the required elements directly inside the circular array.
# The asymptotic complexity remains O(n), but unnecessary
# temporary-list creation and complete rebuilding are avoided.
#
# Before and after execution times were measured using the same
# 10,000-customer workload. The actual times printed by the
# notebook should be recorded in the final performance table.
#
# Conclusion:
# This project demonstrates how data structure selection affects
# the implementation of a practical ticket booking system.
# Both circular queues and linked-list queues support FIFO
# processing efficiently, while their handling of cancellation
# differs. Performance measurement helps identify bottlenecks
# and provides a basis for improving the implementation.
#
# ============================================================
# PERFORMANCE TABLE TO WRITE IN THE NOTEBOOK / REPORT
# ============================================================
#
# | Implementation              | Workload       | Time |
# |-----------------------------|----------------|------|
# | Circular Queue Enqueue      | 10,000 records | Use printed value |
# | Circular Queue Cancellation | 10,000 records | Use printed value |
# | Linked List Enqueue         | 10,000 records | Use printed value |
# | Linked List Cancellation    | 10,000 records | Use printed value |
# | Before Improvement          | 10,000 records | Use printed value |
# | After Improvement           | 10,000 records | Use printed value |
#
# IMPORTANT:
# Do not manually invent timing values.
# Copy the actual values displayed when this notebook is executed.
```
