```python
# Assignment No. 6
# Q1) Implement a hash table using separate chaining.
# Insert the keys 12, 22, 32, 15, 25 and 35 into a table
# of size 10, print the contents of every bucket,
# and report the longest chain length.
#
# a) insert(key, value)
# b) search(key), returning the value or None
# c) delete(key), returning whether the key was present

# Key explanation:
# Separate chaining stores multiple key-value pairs
# in the same bucket using a list.
# The hash index is calculated using key % table size.

class ChainedHashTable:

    def __init__(self, size=10):
        self.size = size
        self.buckets = [[] for _ in range(size)]

    # Calculate bucket index
    def _index(self, key):
        return key % self.size

    # Insert key and value
    def insert(self, key, value):

        index = self._index(key)

        # Update value if key already exists
        for i, (k, v) in enumerate(self.buckets[index]):
            if k == key:
                self.buckets[index][i] = (key, value)
                return

        # Add new key-value pair
        self.buckets[index].append((key, value))

    # Search for a key
    def search(self, key):

        index = self._index(key)

        for k, v in self.buckets[index]:
            if k == key:
                return v

        return None

    # Delete a key
    def delete(self, key):

        index = self._index(key)

        for i, (k, v) in enumerate(self.buckets[index]):
            if k == key:
                del self.buckets[index][i]
                return True

        return False


# Create hash table of size 10
table = ChainedHashTable(10)


# Insert the required keys
table.insert(12, "Value12")
table.insert(22, "Value22")
table.insert(32, "Value32")
table.insert(15, "Value15")
table.insert(25, "Value25")
table.insert(35, "Value35")


# Print every bucket
print("Hash Table Buckets:")

for i, bucket in enumerate(table.buckets):
    print(f"Bucket {i}: {bucket}")


# Find the longest chain
longest_chain = max(
    len(bucket) for bucket in table.buckets
)

print("Longest chain length:", longest_chain)


# Test search operation
print("Search 22:", table.search(22))
print("Search 50:", table.search(50))


# Test delete operation
print("Delete 25:", table.delete(25))
print("Search 25 after deletion:", table.search(25))
```

```python
# Q2) Implement insert and search for a hash table that
# resolves collisions by linear probing.
# Insert 23, 43, 13, 27 and 33 into a table of size 10,
# print the slot each occupies, and explain in two lines
# what problem would arise if a key were later deleted
# by simply emptying its slot.

# Key explanation:
# Linear probing checks the next slot when a collision occurs.
# If the next slot is also occupied, it continues checking
# until an empty slot is found.

class LinearProbingHashTable:

    def __init__(self, size=10):
        self.size = size
        self.table = [None] * size

    # Hash function
    def _hash(self, key):
        return key % self.size

    # Insert a key using linear probing
    def insert(self, key):

        index = self._hash(key)
        start_index = index

        while self.table[index] is not None:

            index = (index + 1) % self.size

            # Table is full
            if index == start_index:
                raise OverflowError("Hash table is full")

        self.table[index] = key

        return index

    # Search for a key
    def search(self, key):

        index = self._hash(key)
        start_index = index

        while self.table[index] is not None:

            if self.table[index] == key:
                return index

            index = (index + 1) % self.size

            if index == start_index:
                break

        return -1


# Create hash table of size 10
table = LinearProbingHashTable(10)


# Required keys
keys = [23, 43, 13, 27, 33]


# Insert keys and print their slots
for key in keys:
    slot = table.insert(key)
    print(f"Key {key} occupies slot {slot}")


# Display the complete table
print("\nHash Table:")

for i, value in enumerate(table.table):
    print(f"Slot {i}: {value}")


# Test search operation
print("\nSearch Results:")

for key in keys:
    slot = table.search(key)
    print(f"Key {key} found at slot {slot}")


# Deletion explanation:
# If a deleted slot is simply made empty, a later search may stop
# at that empty slot and incorrectly report that a key is absent.
# This can break the probing sequence, so deleted slots need a
# special marker such as DELETED.
```

```python
# Q3) Show that the table size matters as much as the hash function.
# Hash the keys 10, 20, 30, 40, 50 and 60 with h(k) = k mod 10
# and again with h(k) = k mod 11.
# Report the number of collisions in each case and explain
# why a prime table size is preferred.

# Key explanation:
# A collision occurs when two or more keys get the same hash index.
# A prime table size can distribute keys more evenly and reduce
# repeated patterns in the hash values.

keys = [10, 20, 30, 40, 50, 60]


# --------------------------------------------------
# Hash function: h(k) = k mod 10
# --------------------------------------------------

table_size_10 = 10
hash_values_10 = {}

collisions_10 = 0

for key in keys:

    index = key % table_size_10

    if index in hash_values_10:
        collisions_10 += 1

    hash_values_10[index] = key


print("Hash function: h(k) = k mod 10")

for key in keys:
    print(f"Key {key} -> Index {key % table_size_10}")

print("Number of collisions:", collisions_10)


# --------------------------------------------------
# Hash function: h(k) = k mod 11
# --------------------------------------------------

table_size_11 = 11
hash_values_11 = {}

collisions_11 = 0

for key in keys:

    index = key % table_size_11

    if index in hash_values_11:
        collisions_11 += 1

    hash_values_11[index] = key


print("\nHash function: h(k) = k mod 11")

for key in keys:
    print(f"Key {key} -> Index {key % table_size_11}")

print("Number of collisions:", collisions_11)


# Explanation:
# With table size 10, all the given keys produce the same
# hash index 0, resulting in 5 collisions.
#
# With table size 11, the keys are distributed across
# different indices, resulting in 0 collisions.
#
# A prime table size is preferred because it helps reduce
# regular patterns and improves the distribution of keys.
```

```python
# Q4) Insert 30 randomly chosen keys into separate-chaining
# tables of size 10, 20 and 50.
# For each table report the longest chain and the average
# chain length, and state how the table size affects the
# cost of a search.

# Key explanation:
# Separate chaining stores collided keys in the same bucket.
# A larger table generally distributes keys across more buckets,
# reducing the average chain length and search cost.

import random


# Generate 30 randomly chosen keys
keys = random.sample(range(1, 1000), 30)


# Function to create a separate-chaining hash table
def create_hash_table(keys, table_size):

    table = [[] for _ in range(table_size)]

    for key in keys:

        index = key % table_size

        table[index].append(key)

    return table


# Table sizes
table_sizes = [10, 20, 50]


# Store results
results = []


for size in table_sizes:

    table = create_hash_table(keys, size)

    # Find longest chain
    longest_chain = max(len(bucket) for bucket in table)

    # Calculate average chain length
    total_elements = sum(len(bucket) for bucket in table)

    average_chain_length = total_elements / size

    results.append(
        (size, longest_chain, average_chain_length)
    )


# Display results
print("Table Size\tLongest Chain\tAverage Chain Length")
print("-" * 60)

for size, longest, average in results:

    print(
        f"{size}\t\t{longest}\t\t{average:.2f}"
    )


# Explanation:
# A smaller table has fewer buckets, so more keys are likely
# to be stored in the same bucket, creating longer chains.
#
# A larger table provides more buckets, which generally reduces
# chain length and makes searching within a bucket faster.
```

```python
# Q5) Count how often each word appears in a short paragraph,
# using your own chaining hash table with the string hash
# from the theory section rather than a Python dictionary.
# Print the three most frequent words.

# Key explanation:
# Separate chaining is used to store words and their frequencies.
# The string hash converts each word into a bucket index.
# The frequency is increased when the same word appears again.

class WordHashTable:

    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    # String hash function from the theory
    def string_hash(self, word):

        total = 0

        for ch in word:
            total += ord(ch)

        return total % self.size

    # Insert word or increase its frequency
    def insert(self, word):

        index = self.string_hash(word)

        for i, (stored_word, count) in enumerate(self.table[index]):

            if stored_word == word:
                self.table[index][i] = (
                    stored_word,
                    count + 1
                )
                return

        self.table[index].append((word, 1))

    # Get all word-frequency pairs
    def get_words(self):

        words = []

        for bucket in self.table:

            for word, count in bucket:
                words.append((word, count))

        return words


# Short paragraph
text = """
data science is useful and data science is powerful
data helps people and science helps people
"""


# Convert text into words
words = text.lower().split()


# Create chaining hash table
table = WordHashTable(10)


# Insert every word
for word in words:
    table.insert(word)


# Get word frequencies
word_counts = table.get_words()


# Sort by frequency in descending order
word_counts.sort(
    key=lambda x: x[1],
    reverse=True
)


# Print three most frequent words
print("Three most frequent words:")

for word, count in word_counts[:3]:
    print(word, "->", count)
```
