```python
# Q1) Implement linear search and binary search on the same
# sorted list of 1000 elements. Count the comparisons each
# makes for a key at the start, middle, end, and a key that
# is absent, and present the results in a table.

# Key explanation:
# Linear search checks elements one by one.
# Binary search repeatedly divides the sorted list into halves.
# Each comparison is counted.

# Create a sorted list of 1000 elements
data = list(range(1000))


# Linear Search
def linear_search(a, key):
    comparisons = 0

    for i in range(len(a)):
        comparisons += 1

        if a[i] == key:
            return i, comparisons

    return -1, comparisons


# Binary Search
def binary_search(a, key):
    low = 0
    high = len(a) - 1
    comparisons = 0

    while low <= high:

        mid = low + (high - low) // 2
        comparisons += 1

        if a[mid] == key:
            return mid, comparisons

        elif a[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1, comparisons


# Keys to be tested
keys = {
    "Start": 0,
    "Middle": 500,
    "End": 999,
    "Absent": 1001
}


# Display comparison table
print("Key Type\tLinear Comparisons\tBinary Comparisons")
print("-" * 60)

for name, key in keys.items():

    linear_index, linear_comparisons = linear_search(data, key)
    binary_index, binary_comparisons = binary_search(data, key)

    print(
        f"{name}\t\t{linear_comparisons}\t\t\t{binary_comparisons}"
    )
```

```python
# Q2) Implement binary search both iteratively and recursively.
# Verify that both return the same index for 100 keys drawn
# from a sorted list, including some keys that are absent.

# Key explanation:
# Binary search works only on a sorted list.
# Both methods repeatedly divide the search range into two halves.
# The iterative method uses a loop, while the recursive method
# calls itself until the key is found or the range becomes empty.

import random


# Sorted list of 1000 elements
data = list(range(1000))


# --------------------------------------------------
# Iterative Binary Search
# --------------------------------------------------

def binary_search_iterative(a, key):

    low = 0
    high = len(a) - 1

    while low <= high:

        mid = low + (high - low) // 2

        if a[mid] == key:
            return mid

        elif a[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# --------------------------------------------------
# Recursive Binary Search
# --------------------------------------------------

def binary_search_recursive(a, key, low, high):

    # Key not found
    if low > high:
        return -1

    mid = low + (high - low) // 2

    if a[mid] == key:
        return mid

    elif a[mid] < key:
        return binary_search_recursive(
            a, key, mid + 1, high
        )

    else:
        return binary_search_recursive(
            a, key, low, mid - 1
        )


# --------------------------------------------------
# Create 100 test keys
# --------------------------------------------------

# 80 keys that exist in the list
existing_keys = random.sample(data, 80)

# 20 keys that are absent from the list
absent_keys = list(range(1000, 1020))

# Combine both types
test_keys = existing_keys + absent_keys

# Shuffle the test keys
random.shuffle(test_keys)


# --------------------------------------------------
# Verify both methods
# --------------------------------------------------

all_same = True

for key in test_keys:

    iterative_result = binary_search_iterative(
        data, key
    )

    recursive_result = binary_search_recursive(
        data, key, 0, len(data) - 1
    )

    # Check whether both methods return the same index
    if iterative_result != recursive_result:
        all_same = False
        print("Mismatch for key:", key)


# Final verification
if all_same:
    print("Both binary search methods returned the same index for all 100 keys.")
else:
    print("Mismatch found.")
```

```python
# Q3) Modify binary search to return the number of comparisons
# it made. Use it to find the largest number of comparisons
# needed for any key in a sorted list of 1000 elements,
# and compare that figure with log2(1000).

# Key explanation:
# Binary search divides the search range into two halves.
# The number of comparisons is counted at every middle-element check.

import math


# Binary Search with comparison counting
def binary_search_count(a, key):

    low = 0
    high = len(a) - 1
    comparisons = 0

    while low <= high:

        mid = low + (high - low) // 2
        comparisons += 1

        if a[mid] == key:
            return mid, comparisons

        elif a[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1, comparisons


# Sorted list of 1000 elements
data = list(range(1000))


# Find the maximum number of comparisons
max_comparisons = 0
max_key = None

for key in data:

    index, comparisons = binary_search_count(data, key)

    if comparisons > max_comparisons:
        max_comparisons = comparisons
        max_key = key


# Calculate log2(1000)
log_value = math.log2(1000)


# Display results
print("Largest number of comparisons:", max_comparisons)
print("Key requiring maximum comparisons:", max_key)
print("log2(1000):", log_value)
```

```python
# Q4 (a) Time a single linear search on an unsorted list
# of 5000 elements.

# Key explanation:
# Linear search checks elements one by one until the key is found.

import random
import time

# Create an unsorted list of 5000 elements
data = list(range(5000))
random.shuffle(data)

# Select a key
key = 4999


def linear_search(a, key):
    for i in range(len(a)):
        if a[i] == key:
            return i
    return -1


# Measure linear search time
start = time.perf_counter()

linear_search(data, key)

linear_time = time.perf_counter() - start

print("Linear search time:", linear_time, "seconds")
```

```python
# Q4 (b) Time the cost of sorting the list and then
# performing one binary search.

# Key explanation:
# Binary search requires a sorted list, so sorting time is
# included along with the binary search time.

import time

# Create another copy of the unsorted list
data_sorted = data.copy()

# Measure sorting time
start = time.perf_counter()

data_sorted.sort()

sorting_time = time.perf_counter() - start


def binary_search(a, key):
    low = 0
    high = len(a) - 1

    while low <= high:

        mid = low + (high - low) // 2

        if a[mid] == key:
            return mid

        elif a[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# Measure one binary search
start = time.perf_counter()

binary_search(data_sorted, key)

binary_time = time.perf_counter() - start

# Total cost
sort_plus_binary_time = sorting_time + binary_time

print("Sorting time:", sorting_time, "seconds")
print("Binary search time:", binary_time, "seconds")
print("Sorting + binary search time:",
      sort_plus_binary_time, "seconds")
```

```python
# Q4 (c) Repeat the comparison for 1000 searches on the same list.

# Key explanation:
# Sorting is performed only once.
# After sorting, all 1000 searches use binary search.

# Generate 1000 search keys
search_keys = [random.randint(0, 4999) for _ in range(1000)]


# --------------------------------------------------
# 1000 Linear Searches
# --------------------------------------------------

start = time.perf_counter()

for key in search_keys:
    linear_search(data, key)

linear_1000_time = time.perf_counter() - start


# --------------------------------------------------
# Sort once + 1000 Binary Searches
# --------------------------------------------------

data_sorted = data.copy()

start = time.perf_counter()

data_sorted.sort()

sorting_time = time.perf_counter() - start


start = time.perf_counter()

for key in search_keys:
    binary_search(data_sorted, key)

binary_1000_time = time.perf_counter() - start


sort_plus_binary_1000_time = (
    sorting_time + binary_1000_time
)


# --------------------------------------------------
# Display comparison
# --------------------------------------------------

print("1000 Linear Searches:",
      linear_1000_time, "seconds")

print("Sorting + 1000 Binary Searches:",
      sort_plus_binary_1000_time, "seconds")


# Determine when sorting pays for itself
if sort_plus_binary_1000_time < linear_1000_time:
    print("Sorting first pays for itself for 1000 searches.")
else:
    print("Sorting first does not pay for itself for 1000 searches.")
```
