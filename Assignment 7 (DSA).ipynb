```python
# Assignment 7
# Q1) Implement a singly linked list class supporting:
# a) insert_at_head, insert_at_tail and insert_at_position
# b) delete_by_value and delete_at_position
# c) search, length and display
# Also test every operation on an empty list,
# a single-element list and a longer list.

# Key explanation:
# Each node stores data and a reference to the next node.
# The head stores the reference to the first node.

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    # Insert at the beginning
    def insert_at_head(self, data):

        node = Node(data)
        node.next = self.head
        self.head = node

    # Insert at the end
    def insert_at_tail(self, data):

        node = Node(data)

        if self.head is None:
            self.head = node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = node

    # Insert at a given position
    def insert_at_position(self, data, pos):

        if pos == 0:
            self.insert_at_head(data)
            return

        current = self.head

        for _ in range(pos - 1):

            if current is None:
                raise IndexError("Position out of range")

            current = current.next

        if current is None:
            raise IndexError("Position out of range")

        node = Node(data)

        node.next = current.next
        current.next = node

    # Delete the first node containing the value
    def delete_by_value(self, data):

        current = self.head
        previous = None

        while current is not None:

            if current.data == data:

                if previous is None:
                    self.head = current.next
                else:
                    previous.next = current.next

                return True

            previous = current
            current = current.next

        return False

    # Delete node at a given position
    def delete_at_position(self, pos):

        if self.head is None:
            return False

        if pos == 0:
            self.head = self.head.next
            return True

        current = self.head

        for _ in range(pos - 1):

            if current.next is None:
                return False

            current = current.next

        if current.next is None:
            return False

        current.next = current.next.next

        return True

    # Search for a value
    def search(self, data):

        current = self.head
        index = 0

        while current is not None:

            if current.data == data:
                return index

            index += 1
            current = current.next

        return -1

    # Find length of linked list
    def length(self):

        count = 0
        current = self.head

        while current is not None:

            count += 1
            current = current.next

        return count

    # Display linked list
    def display(self):

        values = []
        current = self.head

        while current is not None:

            values.append(str(current.data))
            current = current.next

        return " -> ".join(values) + " -> None"


# --------------------------------------------------
# Test 1: Empty list
# --------------------------------------------------

empty_list = SinglyLinkedList()

print("Empty list:")
print(empty_list.display())
print("Length:", empty_list.length())
print("Search 10:", empty_list.search(10))

empty_list.delete_by_value(10)
empty_list.delete_at_position(0)


# --------------------------------------------------
# Test 2: Single-element list
# --------------------------------------------------

single_list = SinglyLinkedList()

single_list.insert_at_head(10)

print("\nSingle-element list:")
print(single_list.display())
print("Length:", single_list.length())
print("Search 10:", single_list.search(10))

single_list.delete_by_value(10)

print("After deletion:")
print(single_list.display())


# --------------------------------------------------
# Test 3: Longer list
# --------------------------------------------------

long_list = SinglyLinkedList()

long_list.insert_at_head(20)
long_list.insert_at_head(10)
long_list.insert_at_tail(40)
long_list.insert_at_position(30, 2)

print("\nLonger list:")
print(long_list.display())
print("Length:", long_list.length())
print("Search 30:", long_list.search(30))

long_list.delete_by_value(20)

print("After deleting value 20:")
print(long_list.display())

long_list.delete_at_position(1)

print("After deleting position 1:")
print(long_list.display())

# Test insertion at position
long_list.insert_at_position(25, 1)

print("After inserting 25 at position 1:")
print(long_list.display())
```

```python
# Q2) Reverse a singly linked list iteratively using the
# three-reference method. Verify it on lists of length
# 0, 1, 2 and 10, and state its time and space complexity.

# Key explanation:
# Three references are used: previous, current and following.
# The next link is reversed one node at a time.

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    def insert_at_tail(self, data):

        node = Node(data)

        if self.head is None:
            self.head = node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = node

    def display(self):

        values = []
        current = self.head

        while current is not None:
            values.append(str(current.data))
            current = current.next

        return " -> ".join(values) + " -> None"

    # Reverse the linked list iteratively
    def reverse(self):

        previous = None
        current = self.head

        while current is not None:

            # Save the next node
            following = current.next

            # Reverse the link
            current.next = previous

            # Move previous forward
            previous = current

            # Move current forward
            current = following

        # Update head
        self.head = previous


# --------------------------------------------------
# Test 1: Empty list
# --------------------------------------------------

list_0 = SinglyLinkedList()

print("Length 0:")
print("Before:", list_0.display())

list_0.reverse()

print("After:", list_0.display())


# --------------------------------------------------
# Test 2: One-element list
# --------------------------------------------------

list_1 = SinglyLinkedList()

list_1.insert_at_tail(10)

print("\nLength 1:")
print("Before:", list_1.display())

list_1.reverse()

print("After:", list_1.display())


# --------------------------------------------------
# Test 3: Two-element list
# --------------------------------------------------

list_2 = SinglyLinkedList()

list_2.insert_at_tail(10)
list_2.insert_at_tail(20)

print("\nLength 2:")
print("Before:", list_2.display())

list_2.reverse()

print("After:", list_2.display())


# --------------------------------------------------
# Test 4: Ten-element list
# --------------------------------------------------

list_10 = SinglyLinkedList()

for i in range(1, 11):
    list_10.insert_at_tail(i)

print("\nLength 10:")
print("Before:", list_10.display())

list_10.reverse()

print("After:", list_10.display())


# Time Complexity: O(n)
# Space Complexity: O(1)
```

```python
# Q3) Add a method that returns the data stored at a given
# position, and use it together with your length() method
# to print the middle element of the list.
# Test it on lists of both odd and even length,
# and state the complexity of your method.

# Key explanation:
# get_at_position() traverses the list until the required position.
# The middle position is found using length() // 2.
# Accessing an element by position takes O(n) time.

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    def insert_at_tail(self, data):

        node = Node(data)

        if self.head is None:
            self.head = node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = node

    def length(self):

        count = 0
        current = self.head

        while current is not None:
            count += 1
            current = current.next

        return count

    # Return data stored at a given position
    def get_at_position(self, pos):

        if pos < 0:
            raise IndexError("Position out of range")

        current = self.head
        index = 0

        while current is not None:

            if index == pos:
                return current.data

            current = current.next
            index += 1

        raise IndexError("Position out of range")


# --------------------------------------------------
# Odd length list
# --------------------------------------------------

odd_list = SinglyLinkedList()

for value in [10, 20, 30, 40, 50]:
    odd_list.insert_at_tail(value)

odd_middle = odd_list.length() // 2

print("Odd length list:")
print("Middle element:", odd_list.get_at_position(odd_middle))


# --------------------------------------------------
# Even length list
# --------------------------------------------------

even_list = SinglyLinkedList()

for value in [10, 20, 30, 40, 50, 60]:
    even_list.insert_at_tail(value)

even_middle = even_list.length() // 2

print("\nEven length list:")
print("Middle element:", even_list.get_at_position(even_middle))


# Complexity:
# Time Complexity: O(n)
# Space Complexity: O(1)
```

```python
# Q4) Compare a linked list with a Python list.
# a) Build both structures with 10000 elements each.
# b) Measure the time to insert 2000 elements at the front of each.
# c) Measure the time to read 2000 elements at random positions
#    in each, then state which structure wins each test and explain why.

# Key explanation:
# Linked lists are efficient for insertion at the front because
# only the head reference needs to be changed.
# Python lists provide O(1) index access, while linked lists
# require traversal to reach a position.

import time
import random


# --------------------------------------------------
# Node class
# --------------------------------------------------

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


# --------------------------------------------------
# Singly Linked List
# --------------------------------------------------

class SinglyLinkedList:

    def __init__(self):
        self.head = None

    def insert_at_head(self, data):

        node = Node(data)

        node.next = self.head
        self.head = node

    def get_at_position(self, pos):

        current = self.head
        index = 0

        while current is not None:

            if index == pos:
                return current.data

            current = current.next
            index += 1

        return None


# --------------------------------------------------
# Build both structures with 10000 elements
# --------------------------------------------------

linked_list = SinglyLinkedList()

for i in range(10000):
    linked_list.insert_at_head(i)


python_list = list(range(10000))


# --------------------------------------------------
# b) Insert 2000 elements at the front
# --------------------------------------------------

# Linked list front insertion
start = time.perf_counter()

for i in range(2000):
    linked_list.insert_at_head(i)

linked_insert_time = time.perf_counter() - start


# Python list front insertion
start = time.perf_counter()

for i in range(2000):
    python_list.insert(0, i)

python_insert_time = time.perf_counter() - start


# --------------------------------------------------
# c) Read 2000 random positions
# --------------------------------------------------

# Generate random positions
random_positions = [
    random.randint(0, 9999)
    for _ in range(2000)
]


# Linked list random access
start = time.perf_counter()

for position in random_positions:
    linked_list.get_at_position(position)

linked_read_time = time.perf_counter() - start


# Python list random access
start = time.perf_counter()

for position in random_positions:
    value = python_list[position]

python_read_time = time.perf_counter() - start


# --------------------------------------------------
# Display results
# --------------------------------------------------

print("Front insertion time:")
print("Linked List:", linked_insert_time, "seconds")
print("Python List:", python_insert_time, "seconds")

print("\nRandom access time:")
print("Linked List:", linked_read_time, "seconds")
print("Python List:", python_read_time, "seconds")


# --------------------------------------------------
# Performance comparison
# --------------------------------------------------

if linked_insert_time < python_insert_time:
    print("\nFront insertion: Linked List is faster.")
else:
    print("\nFront insertion: Python List is faster.")

if linked_read_time < python_read_time:
    print("Random access: Linked List is faster.")
else:
    print("Random access: Python List is faster.")


# Explanation:
# Front insertion is O(1) for a linked list because only the
# head reference is changed.
#
# Front insertion in a Python list is O(n) because existing
# elements must be shifted.
#
# Random access is O(n) for a linked list because nodes must
# be traversed from the head.
#
# Random access is O(1) for a Python list because indexing
# directly accesses the required position.
```

```python
# Assignment 7
# Q5) Write a method that deletes every node holding a given value,
# not only the first, and a second method that removes duplicate
# values from a sorted linked list.
# Verify both, including the case where the value to delete is
# at the head and the case where every node must be removed.

# Key explanation:
# delete_all() removes every node containing the given value.
# remove_duplicates() removes repeated values from a sorted list.

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    # Insert at the end
    def insert_at_tail(self, data):

        node = Node(data)

        if self.head is None:
            self.head = node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = node

    # Display the linked list
    def display(self):

        values = []
        current = self.head

        while current is not None:
            values.append(str(current.data))
            current = current.next

        return " -> ".join(values) + " -> None"

    # Delete every node containing the given value
    def delete_all(self, data):

        # Remove matching nodes from the head
        while self.head is not None and self.head.data == data:
            self.head = self.head.next

        # Remove matching nodes from the remaining list
        current = self.head

        while current is not None and current.next is not None:

            if current.next.data == data:
                current.next = current.next.next
            else:
                current = current.next

    # Remove duplicates from a sorted linked list
    def remove_duplicates(self):

        current = self.head

        while current is not None and current.next is not None:

            if current.data == current.next.data:
                current.next = current.next.next
            else:
                current = current.next


# --------------------------------------------------
# Test 1: Delete every occurrence
# --------------------------------------------------

list1 = SinglyLinkedList()

for value in [10, 20, 20, 30, 20, 40]:
    list1.insert_at_tail(value)

print("Original list:")
print(list1.display())

list1.delete_all(20)

print("After deleting every 20:")
print(list1.display())


# --------------------------------------------------
# Test 2: Delete value at the head
# --------------------------------------------------

list2 = SinglyLinkedList()

for value in [10, 10, 20, 30, 40]:
    list2.insert_at_tail(value)

print("\nBefore deleting head value:")
print(list2.display())

list2.delete_all(10)

print("After deleting every 10:")
print(list2.display())


# --------------------------------------------------
# Test 3: Delete every node
# --------------------------------------------------

list3 = SinglyLinkedList()

for value in [10, 10, 10, 10]:
    list3.insert_at_tail(value)

print("\nBefore deleting every node:")
print(list3.display())

list3.delete_all(10)

print("After deleting every 10:")
print(list3.display())


# --------------------------------------------------
# Test 4: Remove duplicates from sorted list
# --------------------------------------------------

list4 = SinglyLinkedList()

for value in [10, 10, 20, 20, 20, 30, 30, 40]:
    list4.insert_at_tail(value)

print("\nSorted list with duplicates:")
print(list4.display())

list4.remove_duplicates()

print("After removing duplicates:")
print(list4.display())
```
