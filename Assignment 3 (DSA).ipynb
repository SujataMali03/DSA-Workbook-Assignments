```python
# Q1) Implement bubble sort with an early-exit flag.
# Print the array after each pass for the input 5, 1, 4, 2, 8,
# and confirm your passes match the trace given in the theory section.

# Key explanation:
# Bubble sort compares adjacent elements and swaps them
# when they are in the wrong order.
# The swapped flag stops the algorithm if no swap occurs.

def bubble_sort(a):
    n = len(a)

    for i in range(n - 1):

        # Assume no swapping will happen
        swapped = False

        # Compare adjacent elements
        for j in range(n - 1 - i):

            # Swap if elements are in the wrong order
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Print array after each pass
        print("Pass", i + 1, ":", a)

        # Stop if array is already sorted
        if not swapped:
            break

    return a


# Given input
arr = [5, 1, 4, 2, 8]

# Call bubble sort
print("Sorted array:", bubble_sort(arr))
```

```python

# Q2) Implement insertion sort.
# Print the array after each pass for the input 8, 4, 1, 6, 3,
# and report the total number of shift operations performed.

# Key explanation:
# Insertion sort takes one element at a time and inserts it
# into its correct position in the already sorted part.
# A shift occurs when a larger element is moved one position right.

def insertion_sort(a):
    shifts = 0

    # Start from the second element
    for i in range(1, len(a)):

        # Store the current element
        key = a[i]
        j = i - 1

        # Shift larger elements to the right
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
            shifts += 1

        # Insert key into its correct position
        a[j + 1] = key

        # Print array after each pass
        print("Pass", i, ":", a)

    return shifts


# Given input
arr = [8, 4, 1, 6, 3]

# Perform insertion sort
total_shifts = insertion_sort(arr)

# Print final result
print("Sorted array:", arr)
print("Total number of shifts:", total_shifts)



```

```python

# Q3) Compare the two algorithms across different kinds of input.
# a) Generate three arrays of 1000 elements: random, already sorted,
#    and reverse sorted.
# b) Run both algorithms on each, counting comparisons.
# c) Present the results as a 3 × 2 table and identify the
#    best case and worst case for each algorithm.

# Key explanation:
# Comparisons are counted whenever two elements are compared.
# Bubble sort and insertion sort are tested on three different
# types of input to compare their performance.

import random

# Generate arrays of 1000 elements
random_array = [random.randint(1, 10000) for _ in range(1000)]
sorted_array = list(range(1000))
reverse_array = list(range(999, -1, -1))


# Bubble sort with comparison counting
def bubble_counted(a):
    comparisons = 0
    n = len(a)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Stop if the array is already sorted
        if not swapped:
            break

    return comparisons


# Insertion sort with comparison counting
def insertion_counted(a):
    comparisons = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] > key:
                a[j + 1] = a[j]
                j -= 1
            else:
                break

        a[j + 1] = key

    return comparisons


# Store the three types of input
arrays = {
    "Random": random_array,
    "Already Sorted": sorted_array,
    "Reverse Sorted": reverse_array
}


# Compare both algorithms
results = []

for name, data in arrays.items():

    # Use copies so the original data is not changed
    bubble_comparisons = bubble_counted(data.copy())
    insertion_comparisons = insertion_counted(data.copy())

    results.append([
        name,
        bubble_comparisons,
        insertion_comparisons
    ])


# Display the 3 × 2 comparison table
print("Input Type\t\tBubble Sort\tInsertion Sort")

for row in results:
    print(f"{row[0]:<20}\t{row[1]:<12}\t{row[2]}")


# Key observation:
# Best case for both algorithms: Already sorted input.
# Worst case for both algorithms: Reverse sorted input.

```

```python
# Q4) Demonstrate that insertion sort is stable using a list of
# at least six (name, marks) tuples containing repeated marks.
# Then change the comparison in the inner loop from > to >=
# and show that the output order changes.

# Key explanation:
# A stable sorting algorithm keeps the original order of
# elements having equal marks.
# Using > keeps insertion sort stable.

students = [
    ("Asha", 80),
    ("Rahul", 70),
    ("Priya", 80),
    ("Neha", 60),
    ("Amit", 70),
    ("Sneha", 80)
]


def insertion_sort_stable(a):
    # Start from the second element
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        # Use > to maintain stability
        while j >= 0 and a[j][1] > key[1]:
            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = key

    return a


print("Original list:")
print(students)

print("\nUsing > :")
print(insertion_sort_stable(students.copy()))
```

```python
# Key explanation:
# Changing > to >= causes equal-mark elements to shift.
# Therefore, their original order can change.

def insertion_sort_unstable(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        # Changed > to >=
        while j >= 0 and a[j][1] >= key[1]:
            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = key

    return a


print("Using >= :")
print(insertion_sort_unstable(students.copy()))
```

```python
# Q5) Measure both algorithms at n = 500, 1000 and 2000
# on random input, and tabulate the time taken.
# State which was faster at every size, and whether doubling
# n roughly quadrupled the time as O(n²) predicts.

# Key explanation:
# time.perf_counter() is used to measure execution time.
# Both algorithms are tested on the same random input size.

import random
import time


# Bubble Sort
def bubble_sort(a):
    n = len(a)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break


# Insertion Sort
def insertion_sort(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = key


# Test sizes
sizes = [500, 1000, 2000]

# Store results
results = []


for n in sizes:

    # Generate random input
    data = [random.randint(1, 10000) for _ in range(n)]

    # Measure Bubble Sort time
    bubble_data = data.copy()

    start = time.perf_counter()
    bubble_sort(bubble_data)
    bubble_time = time.perf_counter() - start

    # Measure Insertion Sort time
    insertion_data = data.copy()

    start = time.perf_counter()
    insertion_sort(insertion_data)
    insertion_time = time.perf_counter() - start

    # Store results
    results.append((n, bubble_time, insertion_time))


# Display results in a table
print("n\tBubble Sort Time\tInsertion Sort Time")
print("-" * 55)

for n, bubble_time, insertion_time in results:
    print(f"{n}\t{bubble_time:.6f} sec\t\t{insertion_time:.6f} sec")


# Check whether doubling n roughly quadruples the time
print("\nTime Growth Analysis")

for i in range(1, len(results)):
    previous_n, previous_bubble, previous_insertion = results[i - 1]
    current_n, current_bubble, current_insertion = results[i]

    bubble_ratio = current_bubble / previous_bubble
    insertion_ratio = current_insertion / previous_insertion

    print(f"\nFrom n = {previous_n} to n = {current_n}:")
    print(f"Bubble Sort time ratio: {bubble_ratio:.2f}")
    print(f"Insertion Sort time ratio: {insertion_ratio:.2f}")
```
