```python
# Q1) Implement quick sort using the last element as pivot.
# Print the array and the pivot's final index after every
# partitioning step for the input 8, 3, 9, 2, 7.

# Key explanation:
# The last element is selected as the pivot.
# Elements smaller than the pivot are placed on the left
# and larger elements are placed on the right.
# After partitioning, the pivot reaches its final position.

def partition(a, low, high):
    # Select the last element as pivot
    pivot = a[high]

    # Index of the smaller element
    i = low - 1

    # Compare each element with the pivot
    for j in range(low, high):
        if a[j] < pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    # Place pivot in its final position
    a[i + 1], a[high] = a[high], a[i + 1]

    # Return pivot's final index
    return i + 1


def quick_sort(a, low, high):
    if low < high:

        # Partition the array
        p = partition(a, low, high)

        # Print array and pivot index
        print("Array:", a)
        print("Pivot final index:", p)

        # Sort left part
        quick_sort(a, low, p - 1)

        # Sort right part
        quick_sort(a, p + 1, high)


# Given input
arr = [8, 3, 9, 2, 7]

# Apply quick sort
quick_sort(arr, 0, len(arr) - 1)

# Print final sorted array
print("Sorted array:", arr)
```

```python
# Q3) Show that quick sort's performance depends on the input.
# a) Sort an already-sorted array of 500 elements with a
#    last-element pivot, and record the recursion depth reached.
# b) Sort a random array of the same size and record the depth again.
# c) Explain in two lines why the two depths differ so sharply.

# Key explanation:
# The last element is used as the pivot.
# Recursion depth is counted for every recursive Quick Sort call.

import random

def partition(a, low, high):
    # Select the last element as pivot
    pivot = a[high]

    i = low - 1

    # Place smaller elements before the pivot
    for j in range(low, high):
        if a[j] < pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    # Place pivot in its correct position
    a[i + 1], a[high] = a[high], a[i + 1]

    return i + 1


def quick_sort_depth(a, low, high, depth=1):
    # Return maximum recursion depth
    if low >= high:
        return depth

    p = partition(a, low, high)

    left_depth = quick_sort_depth(a, low, p - 1, depth + 1)
    right_depth = quick_sort_depth(a, p + 1, high, depth + 1)

    return max(left_depth, right_depth)


# --------------------------------------------------
# (a) Already-sorted array of 500 elements
# --------------------------------------------------

sorted_array = list(range(500))

sorted_depth = quick_sort_depth(
    sorted_array,
    0,
    len(sorted_array) - 1
)

print("Recursion depth for sorted array:", sorted_depth)


# --------------------------------------------------
# (b) Random array of 500 elements
# --------------------------------------------------

random_array = list(range(500))
random.shuffle(random_array)

random_depth = quick_sort_depth(
    random_array,
    0,
    len(random_array) - 1
)

print("Recursion depth for random array:", random_depth)


# --------------------------------------------------
# (c) Explanation
# --------------------------------------------------

# For a sorted array, the last element is always the largest,
# so the partition becomes highly unbalanced and recursion depth
# becomes very large.
#
# For a random array, partitions are generally more balanced,
# so the recursion depth is much smaller.
```

```python
# Q4) Run bubble sort, insertion sort and quick sort on random
# arrays of size 500, 1000 and 2000, counting comparisons
# for each. Present a 3 × 3 table.

# Key explanation:
# A comparison is counted whenever two elements are compared.
# The same random input is used for all three algorithms
# at each array size.

import random


# --------------------------------------------------
# Bubble Sort
# --------------------------------------------------

def bubble_counted(a):
    comparisons = 0
    n = len(a)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break

    return comparisons


# --------------------------------------------------
# Insertion Sort
# --------------------------------------------------

def insertion_counted(a):
    comparisons = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] > key:
                a[j + 1] = a[j]
                j -= 1
            else:
                break

        a[j + 1] = key

    return comparisons


# --------------------------------------------------
# Quick Sort
# --------------------------------------------------

def quick_counted(a, low, high, stats):

    if low < high:

        pivot = a[high]
        i = low - 1

        for j in range(low, high):

            stats["comparisons"] += 1

            if a[j] < pivot:
                i += 1
                a[i], a[j] = a[j], a[i]

        # Place pivot in correct position
        a[i + 1], a[high] = a[high], a[i + 1]

        p = i + 1

        # Sort left part
        quick_counted(a, low, p - 1, stats)

        # Sort right part
        quick_counted(a, p + 1, high, stats)


# --------------------------------------------------
# Test for n = 500, 1000 and 2000
# --------------------------------------------------

sizes = [500, 1000, 2000]

results = []

for n in sizes:

    # Generate random array
    data = [random.randint(1, 10000) for _ in range(n)]

    # Bubble Sort comparisons
    bubble_comparisons = bubble_counted(data.copy())

    # Insertion Sort comparisons
    insertion_comparisons = insertion_counted(data.copy())

    # Quick Sort comparisons
    quick_data = data.copy()
    stats = {"comparisons": 0}

    quick_counted(
        quick_data,
        0,
        len(quick_data) - 1,
        stats
    )

    quick_comparisons = stats["comparisons"]

    # Store results
    results.append(
        (n, bubble_comparisons,
         insertion_comparisons,
         quick_comparisons)
    )


# --------------------------------------------------
# Display 3 × 3 comparison table
# --------------------------------------------------

print("n\tBubble Sort\tInsertion Sort\tQuick Sort")
print("-" * 60)

for n, bubble, insertion, quick in results:
    print(f"{n}\t{bubble}\t\t{insertion}\t\t{quick}")


# Identify the algorithm with fewest comparisons
for n, bubble, insertion, quick in results:

    values = {
        "Bubble Sort": bubble,
        "Insertion Sort": insertion,
        "Quick Sort": quick
    }

    fewest = min(values, key=values.get)

    print(f"\nFor n = {n}, fewest comparisons: {fewest}")
```

```python
# Q5) Construct an input that forces your quick sort into its O(n²)
# worst case, and demonstrate it by measuring the time for that
# input against a random input of the same size.
# Explain why that particular input is the worst case for
# a last-element pivot.

# Key explanation:
# An already sorted array is a worst-case input when the last
# element is selected as the pivot.
# The pivot becomes the largest element, so one partition
# contains n-1 elements and the other contains 0 elements.
# This creates highly unbalanced partitions and O(n²) time.

import random
import time


# Partition function
def partition(a, low, high):
    pivot = a[high]
    i = low - 1

    for j in range(low, high):
        if a[j] < pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    a[i + 1], a[high] = a[high], a[i + 1]

    return i + 1


# Quick Sort
def quick_sort(a, low, high):
    if low < high:

        p = partition(a, low, high)

        quick_sort(a, low, p - 1)
        quick_sort(a, p + 1, high)


# Size of input
n = 1000


# --------------------------------------------------
# Worst-case input
# --------------------------------------------------

# Already sorted input forces the worst case
worst_case = list(range(n))


# Measure worst-case time
start = time.perf_counter()

quick_sort(worst_case, 0, len(worst_case) - 1)

worst_time = time.perf_counter() - start


# --------------------------------------------------
# Random input
# --------------------------------------------------

random_input = list(range(n))
random.shuffle(random_input)


# Measure random-input time
start = time.perf_counter()

quick_sort(random_input, 0, len(random_input) - 1)

random_time = time.perf_counter() - start


# Display results
print("Worst-case input time:", worst_time, "seconds")
print("Random input time:", random_time, "seconds")


# Explanation:
# For the sorted array, the last element is always the largest.
# Therefore, every partition produces one large subarray and
# one empty subarray, causing O(n²) performance.
```
