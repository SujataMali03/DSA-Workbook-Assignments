```python
#ASSIGNMENT NO. 1: Introduction to Algorithms, Data Structures and
#Complexity Analysis
#Q1) Write a program that finds the largest and the smallest value in a list of numbers without using
#max() or min(). Report how many comparisons your code performed and state its time complexity in
#Big-O notation
numbers = [12, 5, 8, 20, 3, 15]

largest = numbers[0]
smallest = numbers[0]
comparisons = 0

for num in numbers[1:]:
    comparisons += 1
    if num > largest:
        largest = num

    comparisons += 1
    if num < smallest:
        smallest = num

print("Largest value:", largest)
print("Smallest value:", smallest)
print("Number of comparisons:", comparisons)
```

```python
### Q3. Investigate how running time grows with input size.

#**a) Write three functions on a list of size n: one O(1), one O(n) and one O(n²).
#b) Time each at n = 500, 1000 and 2000, and present the results as a table of n against time.
#c) State which function grew fastest, and whether that matches its Big-O complexity.**

### Answer:


import time

# O(1)
def constant_time(a):
    return a[0]

# O(n)
def linear_time(a):
    total = 0
    for x in a:
        total += x
    return total

# O(n²)
def quadratic_time(a):
    count = 0
    for i in range(len(a)):
        for j in range(i + 1, len(a)):
            count += 1
    return count

sizes = [500, 1000, 2000]

for n in sizes:
    data = list(range(n))

    start = time.perf_counter()
    constant_time(data)
    t1 = time.perf_counter() - start

    start = time.perf_counter()
    linear_time(data)
    t2 = time.perf_counter() - start

    start = time.perf_counter()
    quadratic_time(data)
    t3 = time.perf_counter() - start

    print(n, round(t1, 6), round(t2, 6), round(t3, 6))
```

```python
### Q4. Count the comparisons a linear search makes on a list of 1000 elements for a key that is the first element,
#a key in the middle, and a key that is absent. Identify which of the three is the best case and which is the worst case,
#and explain why.

### Answer:

def linear_search(a, key):
    comparisons = 0

    for x in a:
        comparisons += 1

        if x == key:
            return comparisons

    return comparisons


data = list(range(1, 1001))

first = linear_search(data, 1)
middle = linear_search(data, 501)
absent = linear_search(data, 1001)

print("First element comparisons:", first)
print("Middle element comparisons:", middle)
print("Absent key comparisons:", absent)
```

```python
### Q5. Compute the sum 1 + 2 + … + n in two ways: with a loop, and with the formula n × (n + 1) / 2. 
#Verify that both give the same answer for n = 1 to 100, time both at n = 1000000, and state the complexity of each.
#Explain what this shows about the value of choosing a better algorithm.

### Answer:


import time

# Using a loop
def loop_sum(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total


# Using the formula
def formula_sum(n):
    return n * (n + 1) // 2


# Verify both methods for n = 1 to 100
for n in range(1, 101):
    assert loop_sum(n) == formula_sum(n)


# Time both methods for n = 1000000
n = 1000000

start = time.perf_counter()
loop_sum(n)
loop_time = time.perf_counter() - start

start = time.perf_counter()
formula_sum(n)
formula_time = time.perf_counter() - start

print("Loop time:", loop_time)
print("Formula time:", formula_time)
```
