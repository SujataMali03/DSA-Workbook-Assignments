```python
# Q1. Implement the following three functions recursively, stating the base case of each.

#a) The sum of the first n natural numbers.
#b) The factorial of n.
#c) The sum of all elements in a list.

# Answer:


# a) Sum of first n natural numbers
def sum_n(n):
    if n == 0:          # Base case
        return 0
    return n + sum_n(n - 1)


# b) Factorial of n
def factorial(n):
    if n == 0:          # Base case
        return 1
    return n * factorial(n - 1)


# c) Sum of all elements in a list
def list_sum(a, n):
    if n == 0:          # Base case
        return 0
    return a[n - 1] + list_sum(a, n - 1)

```

```python
# Q2) Write a recursive function that counts the number of digits in a positive integer, and another that reverses a string. Test both on at least five inputs, including the smallest valid input for each.

# Answer:

#1. Recursive function to count the number of digits

# Function to count the number of digits in a number
def count_digits(n):

    # Base case: if number has only one digit
    if n < 10:
        return 1

    # Remove the last digit and count again
    return 1 + count_digits(n // 10)


# Five test inputs
numbers = [1, 5, 12, 123, 1000]

# Call the function for each number
for n in numbers:
    print(count_digits(n))


#Base case: n < 10` because a single-digit number contains exactly one digit.

#2. Recursive function to reverse a string**

# Function to reverse a string
def reverse_string(s):

    # Base case: string has one or zero characters
    if len(s) <= 1:
        return s

    # Reverse the remaining string and add first character at the end
    return reverse_string(s[1:]) + s[0]


# Five test inputs
strings = ["a", "hi", "cat", "hello", "python"]

# Call the function for each string
for s in strings:
    print(reverse_string(s))


#Base case: `len(s) <= 1` because a string with one or zero characters is already reversed.
```

```python
# Q3) Compute the nth Fibonacci number recursively, with a counter for the number of calls made. Report the call count for n = 5,
#10 and 15 in a table, and explain, using the recursion tree, why the count grows so quickly.
# Answer:

# Counter to keep track of the number of function calls
calls = 0

# Recursive Fibonacci function
def fib_counted(n):

    # Use the global counter variable
    global calls
    calls += 1

    # Base case: Fibonacci of 0 or 1
    if n <= 1:
        return n

    # Recursive case: calculate two previous Fibonacci numbers
    return fib_counted(n - 1) + fib_counted(n - 2)


# Test for n = 5, 10 and 15
for n in [5, 10, 15]:

    # Reset counter for each value of n
    calls = 0

    # Calculate Fibonacci number
    fib_counted(n)

    # Display the number of calls
    print(n, calls)


#Call Count Table:

#| n  | Number of Calls |
#| -- | --------------: |
#| 5  |              15 |
#| 10 |             177 |
#| 15 |            1973 |

#Explanation:
#For every value greater than 1, the function makes **two recursive calls**: `fib(n-1)` and `fib(n-2)`.
#These calls again create more calls, forming a large **recursion tree**.
#Many of the same Fibonacci values are calculated repeatedly, so the number of function calls grows very quickly, approximately exponentially.
```

```python
# Q4) Write a recursive function that counts down from n to 0. Find by trial the largest n it can handle before Python raises RecursionError,
#report that value, explain in one line what is running out, and rewrite the same countdown iteratively so that no limit is reached.
# Answer:
#1. Recursive Countdown
# Recursive function to count down from n to 0
def recursive_countdown(n):

    # Base case: stop when n becomes 0
    if n == 0:
        print(0)
        return

    # Print the current number
    print(n)

    # Recursive call with n decreased by 1
    recursive_countdown(n - 1)


# Find the largest n before RecursionError
n = 1

while True:
    try:
        recursive_countdown(n)
        n += 1
    except RecursionError:
        print("Recursion limit reached at:", n)
        break


#Largest n: It should be found by trial on the system; Python's default recursion limit is approximately **1000**.

#What runs out: The **recursion call stack** runs out of available stack depth.

#2. Iterative Countdown


# Iterative function to count down from n to 0
def iterative_countdown(n):

    # Continue until n becomes 0
    while n >= 0:

        # Print the current number
        print(n)

        # Decrease n by 1
        n -= 1


#Advantage: The iterative version does not create a new recursive call for every number, so it does not face Python's recursion-depth limit.
```

```python
### Q5) Compute x raised to the power n in two recursive ways: by multiplying x by itself n times, and by halving the exponent as shown in the theory section. Count the multiplications each performs for n = 16 and n = 100, and explain the difference.

### Answer:

#1. Power using simple recursion


# Function to calculate x raised to power n
def power_slow(x, n):

    # Base case: any number raised to power 0 is 1
    if n == 0:
        return 1

    # Multiply x once and reduce the power by 1
    return x * power_slow(x, n - 1)


#This method performs n multiplications.


#2. Power using exponent halving


# Function to calculate power using exponent halving
def power_fast(x, n):

    # Base case: any number raised to power 0 is 1
    if n == 0:
        return 1

    # Calculate x raised to half of n
    half = power_fast(x, n // 2)

    # If n is even, multiply the half result by itself
    if n % 2 == 0:
        return half * half

    # If n is odd, multiply x with the two half results
    return x * half * half



#Explanation:
#The simple recursive method reduces `n` by **1**, so it needs **O(n)** multiplications.
#The exponent-halving method reduces `n` by **half** at each step,
#so it needs approximately **O(log n)** multiplications and is much faster for large values of `n`.
```
